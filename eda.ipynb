# Exploratory Data Analysis (EDA) for Base.csv
This notebook provides a quick overview of the dataset used in `run.py`.

```python
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid", context="notebook")
%matplotlib inline
```

```python
# Load dataset
csv_path = "Base.csv"
if not os.path.exists(csv_path):
    raise FileNotFoundError("Base.csv not found. Run run.py first to download/create it.")

df = pd.read_csv(csv_path)
df.head()
```

```python
# Basic overview
print("Shape:", df.shape)
display(df.head(10))
print("\nInfo:")
df.info()

print("\nNumeric summary:")
display(df.describe(include=[np.number]).T)

print("\nCategorical summary:")
display(df.describe(include=["object", "category"]).T)
```

```python
# Missing values
missing = df.isna().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(2)
miss_df = pd.DataFrame({"missing": missing, "percent": missing_pct})
display(miss_df[miss_df["missing"] > 0])
```

```python
# Target distribution (if present)
target_col = "fraud_bool" if "fraud_bool" in df.columns else None
if target_col is not None:
    counts = df[target_col].value_counts(dropna=False)
    pct = (counts / len(df) * 100).round(2)
    disp = pd.DataFrame({"count": counts, "percent": pct})
    display(disp)
    ax = sns.countplot(x=target_col, data=df)
    ax.set_title("Target distribution")
    plt.show()
else:
    print("Target column 'fraud_bool' not found.")
```

```python
# Categorical features overview (predefined set from run.py)
categorical_features = [
    "payment_type",
    "employment_status",
    "housing_status",
    "source",
    "device_os",
]
cats_present = [c for c in categorical_features if c in df.columns]
if cats_present:
    nunique = df[cats_present].nunique().sort_values(ascending=False)
    display(pd.DataFrame({"nunique": nunique}))
    for c in cats_present:
        top = df[c].value_counts().head(10)
        ax = sns.countplot(y=c, data=df, order=top.index)
        ax.set_title(f"Top categories for {c}")
        plt.show()
else:
    print("No predefined categorical features present.")
```

```python
# Numeric feature distributions (subset)
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
num_cols = [c for c in num_cols if c != (target_col or "")]
n = min(12, len(num_cols))
if n > 0:
    cols = 3
    rows = int(np.ceil(n / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(5 * cols, 3 * rows))
    axes = axes.ravel() if isinstance(axes, np.ndarray) else [axes]
    for i, c in enumerate(num_cols[:n]):
        sns.histplot(df[c].dropna(), bins=30, ax=axes[i], kde=False)
        axes[i].set_title(c)
    for j in range(i + 1, rows * cols):
        fig.delaxes(axes[j])
    fig.suptitle("Numeric feature distributions (subset)", y=1.02)
    plt.tight_layout()
    plt.show()
else:
    print("No numeric features found.")
```

```python
# Correlation heatmap and correlation with target
if len(num_cols) > 1:
    cols_for_corr = num_cols.copy()
    if target_col and target_col not in cols_for_corr and df[target_col].dtype.kind in 'biufc':
        cols_for_corr.append(target_col)
    corr = df[cols_for_corr].corr(numeric_only=True)
    plt.figure(figsize=(10, 8))
    sns.heatmap(corr, cmap="coolwarm", center=0)
    plt.title("Correlation heatmap")
    plt.show()
    if target_col and target_col in corr.columns:
        tgt_corr = corr[target_col].drop(target_col, errors='ignore').sort_values(ascending=False)
        display(pd.DataFrame({"corr_with_target": tgt_corr}))
else:
    print("Not enough numeric features for correlation.")
```
