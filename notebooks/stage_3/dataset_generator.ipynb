```python
from sklearn.datasets import make_classification
import pandas as pd

X, y = make_classification(
    n_samples=1500,
    n_features=10,
    n_informative=6,
    n_redundant=2,
    n_repeated=0,
    n_classes=2,
    n_clusters_per_class=2,
    class_sep=1.5,
    weights=[0.6, 0.4],
    flip_y=0.02,
    random_state=42
)

df = pd.DataFrame(X, columns=[f'f{i+1}' for i in range(X.shape[1])])
df['target'] = y

df.to_csv('../../datasets/synthetic_classification_data.csv', index=False)
```
